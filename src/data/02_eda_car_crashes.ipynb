## Correlation Analysis Summary

### Strong correlations with total_deaths:
- It was observed that total_deaths has a higher and positive correlation with alcohol and speeding. Meaning that, when the number of people that are under the influence of alcohol or speeding, the number of total_deaths increases.

### Spurious correlations (and why):
- not_distracted and no_previous. There's some positive correlation between these two features and total_deaths, however, we cannot assume cause.
- Logically speaking, it is statically correct saying that most of the people that were killed driving was not distracted.
- One of the most curious correlation is the "no_previous": Let us think if there's any relationship anyhow in the number of people that didn't have any previous accidents and the total number of deaths. However, it is also statistically correct, there's a low ratio of people in a more generic universe (statistically speaking) that suffered an accident.

### Feature independence:
- we can see some statistically correlation and cause (also logical) between total_deaths x speeding, total_deaths x alcohol and speeding_alcohol

### Dropped/kept decisions:
- Considering that, there's no signigicant relationship between ins_losses and total_death, ins_premium and total_death. We should drop these two columns.
- It is indeed intriguing in why still we have a -0.19 value for PCC calculations between ins_premium vs total_death. Could lead to some demographic assumption, but hard to tell with so low number of data rows.

```python
from sklearn.model_selection import train_test_split
```

```python
x = df_car_crashes[['alcohol', 'speeding']]
y = df_car_crashes['total_deaths']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(x_train.head())
print(f"""Number of train records: {len(x_train)}""")

print(y_test.head())
print(f"""Number of test records: {len(y_test)}""")
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
est = lr.fit(x_train, y_train)
```

```python
print(est.coef_)
print(est.intercept_)

print(f"""If alcohol goes up by 1 and speeding maintains constant (0): """)
print(f"""y = w1 * alcohol + w2 * speeding + b => {(est.coef_[0] * 1) + est.intercept_}""")

print(f"""If speeding goes up by 1 and alcohol maintains constant (0): """)
print(f"""y = w1 * alcohol + w2 * speeding + b => {(est.coef_[1] * 1) + est.intercept_}""")
```

```python
y_pred = lr.predict(x_test)

for actual, predicted in zip(y_test.values, y_pred):
    print(f"{actual} => {predicted}")
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test, y_pred)
print(f"""MSE: {mse}""")
```

```python
import math
print(f"RMSE: {math.sqrt(mse):.2f}")
```
